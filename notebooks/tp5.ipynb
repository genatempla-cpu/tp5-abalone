La reproducibilidad asegura que, dadas las mismas entradas y condiciones, el resultado del análisis o la construcción del
modelo será idéntico

Los Tres Pilares de un Proyecto Trazable: Entornos, Código y Datos:

Codigo: Es el pilar fundamental del versionado y se maneja típicamente mediante sistemas de control
de versiones distribuidos (DVCS) como Git. Git permite a los desarrolladores registrar cambios al repositorio,
ver el historial de commits, deshacer acciones, y gestionar flujos de trabajo de desarrollo complejos a través de ramas(branches).

Entornos: Esto se refiere a la configuración del hardware y software en la que se ejecuta el código.
En ML, la plataforma (el 95% del sistema de producción) incluye la gestión de recursos. Para la reproducibili
dad, es crucial que los entornos de ejecución sean consistentes. La contenerización, utilizando tecnologías como
Docker, permite empaquetar la implementación del modelo y sus dependencias.

Datos: Los datos son el combustible del ML, y su versionado es tan importante como el del código. Un
modelo entrenado en un conjunto de datos específico debe poder recrearse con precisión utilizando exactamente
la misma versión de los datos.
